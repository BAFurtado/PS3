## Data

Set `SAVE = True` to also write the PDF, PNG and CSVs to `F.OUT`.

```python
ROOT = 'output/emissions_calibra4/lam_20.0_mu_0.992_step_0.995_tax_0.2'   # later: 'output/emissions'
SAVE = False
batch = data.load(ROOT)
print('scenarios:', batch.scenarios, '| TAX pairs:', len(batch.pairs('TAX')),
      '| capitals:', batch.stats.city.nunique(), '| agents saved:', not batch.workers.empty)
```

```python
def show(name, fig, numbers):
    if SAVE:
        F.save(fig, numbers, name)
    return numbers
```
